# Enterprise Content AI Prototype

This prototype demonstrates a **multi-agent content workflow** for enterprise operations, including:

- Drafting content from a user prompt (**Knowledge-to-content agent**)  
- Reviewing for **brand tone, terminology, and compliance** (**Brand governance agent**)  
- **Localization** for different languages  
- **Multi-channel distribution** (simulated)  
- **Analytics and engagement feedback** (**Content intelligence agent**)  
- **Content version tracking**  

This notebook showcases a **full lifecycle automation workflow** suitable for enterprise content operations.

```python
!pip install --upgrade google-generativeai gradio

import os
from google.colab import userdata
import google.genai as genai
import random

# -----------------------
# Setup Gemini API
# -----------------------
os.environ["GEMINI_API_KEY"] = userdata.get("GEMINI_API_KEY")
client = genai.Client(api_key=os.environ["GEMINI_API_KEY"])
```

## Instructions

1. Enter your **content prompt** in the UI  
2. The system will:
   - Generate a draft  
   - Review for compliance  
   - Localize content  
   - Simulate distribution  
   - Provide analytics + metrics  

```python
# -----------------------
# Multi-Agent Pipeline Functions
# -----------------------

def safe_generate(prompt, retries=3):
    """Handles API failures safely."""
    for i in range(retries):
        try:
            response = client.models.generate_content(
                model="gemini-2.5-flash",
                contents=prompt
            )
            return response.text
        except Exception as e:
            if i == retries - 1:
                return f"Error: {str(e)}"


def draft_content(user_prompt):
    return safe_generate(
        f"Create professional enterprise content based on this prompt:\n{user_prompt}"
    )


def review_content(draft_text):
    return safe_generate(f"""
Check the following content for:
- Brand tone
- Terminology consistency
- Regulatory/legal compliance

Improve and return final version:

{draft_text}
""")


# ✅ FIXED (important)
def localize_content(text, target_language="English"):
    if target_language == "English":
        return text

    return safe_generate(f"""
Translate and adapt the following content to {target_language}.
Maintain professional tone.

{text}
""")


def distribute_content(text):
    channels = ["Website", "LinkedIn", "Twitter", "Email"]
    return "\n".join([f"Published to {c}" for c in channels])


def content_analytics(text):
    return safe_generate(f"""
Analyze this content and suggest improvements for:
- Engagement
- Tone
- Clarity
- Audience targeting

{text}
""")


def simulate_metrics():
    return {
        "Views": random.randint(500, 5000),
        "Clicks": random.randint(50, 500),
        "Shares": random.randint(10, 100)
    }
```

## Interactive AI Content System

Use the interface below to generate, review, localize, and analyze content using a multi-agent AI pipeline.

```python
import gradio as gr
import matplotlib.pyplot as plt

def run_pipeline(user_prompt, language):
    try:
        # 1. Draft
        draft = draft_content(user_prompt)

        # 2. Review
        reviewed = review_content(draft)

        # 3. Auto approval
        approved = reviewed

        # 4. Localization
        localized = localize_content(approved, language)

        # 5. Distribution
        distribution = distribute_content(localized)

        # 6. Analytics
        feedback = content_analytics(localized)

        # 7. Metrics
        metrics = simulate_metrics()

        # Chart
        plt.figure()
        plt.bar(list(metrics.keys()), list(metrics.values()))
        plt.title("Engagement Metrics")
        plt.xlabel("Metric")
        plt.ylabel("Value")

        return draft, reviewed, localized, distribution, feedback, str(metrics), plt.gcf()

    except Exception as e:
        return "Error", "Error", "Error", "Error", str(e), "Error", None


with gr.Blocks(theme=gr.themes.Soft()) as demo:

    gr.Markdown("# 🚀 Enterprise Content AI System")
    gr.Markdown("AI-powered multi-agent content pipeline")

    with gr.Row():
        prompt = gr.Textbox(
            label="Enter Content Prompt",
            lines=3
        )

        language = gr.Dropdown(
            [
                "English",
                "Spanish",
                "French",
                "German",
                "Hindi",
                "Chinese (Simplified)",
                "Arabic",
                "Portuguese",
                "Japanese",
                "Russian"
            ],
            value="English",
            label="Select Language"
        )

    run_btn = gr.Button("Generate Content")

    gr.Markdown("## ✍️ Draft")
    draft_out = gr.Textbox(lines=5)

    gr.Markdown("## ✅ Reviewed")
    review_out = gr.Textbox(lines=5)

    gr.Markdown("## 🌍 Localized")
    localized_out = gr.Textbox(lines=5)

    gr.Markdown("## 📢 Distribution")
    dist_out = gr.Textbox(lines=3)

    gr.Markdown("## 📊 Analytics")
    feedback_out = gr.Textbox(lines=5)

    gr.Markdown("## 📈 Metrics")
    metrics_out = gr.Textbox()
    chart_out = gr.Plot()

    run_btn.click(
        run_pipeline,
        inputs=[prompt, language],
        outputs=[
            draft_out,
            review_out,
            localized_out,
            dist_out,
            feedback_out,
            metrics_out,
            chart_out
        ]
    )

demo.launch()
```
